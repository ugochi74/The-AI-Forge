This notebook compares a Python list with a NumPy array, reads a small CSV, and plots model accuracy. Run it from top to bottom. The kernel keeps variables in memory between cells.

## 1. Load the tools and lesson data

The CSV is a small fixture stored with this lesson. These imports belong in a code cell; this explanation belongs in a Markdown cell.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

lesson_dir = Path.cwd()
if not (lesson_dir / 'outputs/model_metrics.csv').exists():
    lesson_dir = Path('phases/00-setup-and-tooling/05-jupyter-notebooks')
csv_path = lesson_dir / 'outputs/model_metrics.csv'
df = pd.read_csv(csv_path)
df
```

## 2. Plot the accuracy values

The plot is useful during exploration because it appears beside the code and data.

```python
%matplotlib inline
ax = df.plot(x='model', y='accuracy', kind='bar', legend=False)
ax.set_title('Model accuracy')
ax.set_ylabel('Accuracy')
ax.set_ylim(0, 1)
plt.tight_layout()
plt.show()
```

## 3. Compare timings

`%timeit` repeats a single line and reports an average. The two lines do the same kind of work: generate 100,000 random numbers.

```python
import random
import numpy as np
random.seed(42)
%timeit [random.random() for _ in range(100_000)]
%timeit np.random.random(100_000)
```

`%%time` runs an entire cell once. Use it for a longer operation where repeating the work many times would be unhelpful.

```python
%%time
total = sum(range(1_000_000))
print(f'Total: {total:,}')
```
